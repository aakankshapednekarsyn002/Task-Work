```python
# Duplicate values Check
duplicates = df.duplicated().sum()
print("Duplicate Row Count: ")
print(duplicates)
```

```python
df['sales'] = df['units_sold'] * df['price_unit']
df.head()
```

```python
# Hstogram for Sales Distribution
plt.figure(figsize=(10,6))
sns.histplot(df['sales'], bins=50)
plt.title("Sales Distribution")
plt.show()
```

Most sales values are in the lower range, while a few transactions have higher values. This shows that most products are sold in moderate amounts.

```python
plt.figure(figsize=(10,6))
df.groupby('sku')['sales'].sum().sort_values(ascending=False).head(10).plot(kind='bar', color="darkblue")
plt.title("Top 10 Products by Sales")
plt.xlabel("Product (SKU)")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Here are Top 10 products, which generates most of the sales. The company should focus on these high-performing products to increase revenue.

```python
plt.figure(figsize=(10,6))
df.groupby('category')['sales'].sum().sort_values(ascending=False).plot(kind='bar', color="red")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Some categories contribute more to total sales than others. The business should invest more in high-performing categories.

```python
plt.figure(figsize=(8,5))
df.groupby('promotion_flag')['sales'].mean().plot(kind='bar', color="purple")
plt.title("Average Sales: Promotion vs No Promotion")
plt.xlabel("Promotion Flag")
plt.ylabel("Average Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Promotions increase sales on average, showing that discounts and offers are effective in boosting demand.

```python
plt.figure(figsize=(10,6))
df.groupby('region')['sales'].sum().sort_values(ascending=False).plot(kind='bar', color="salmon")
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

These regions perform better than others. The company can expand marketing and supply in these high-demand regions.

```python
df['date'] = pd.to_datetime(df['date'])
df['Month'] = df['date'].dt.to_period('M')

plt.figure(figsize=(10,6))
df.groupby('Month')['sales'].sum().plot(kind='line', color="darkblue")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

Sales change over time, showing trends that can help in planning inventory and promotions.
