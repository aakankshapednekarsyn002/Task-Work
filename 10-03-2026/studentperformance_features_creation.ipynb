```python
# Importing necessary libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
```

```python
# Load the Data
df = pd.read_csv("StudentPerformanceFactors.csv")
```

```python
# First 5 rows
df.head()
```

```python
# Last 5 rows
df.tail()
```

```python
print("Shape of the Data: ")
print(df.shape)
```

```python
print("Summary of the dataset: ")
print(df.info())
```

```python
# Standardise the Column Names
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
)
print(df.columns)
```

```python
# Datatypes of Each Column
print("Datatypes of Each Column :")
print(df.dtypes)
```

```python
# Numerical Columns Statistical Summary
print("Statistic Summary for Numerical Columns:")
print(df.describe())
```

```python
# Missing Values check
missing_values = df.isnull().sum()
print("Missing Values for Each Column")
print(missing_values)
```

```python
# Duplicate values Check
duplicates = df.duplicated().sum()
print("Duplicate Row Count: ")
print(duplicates)
```

```python
# Handle Missing Values
df['teacher_quality'] = df['teacher_quality'].fillna('Unknown')
df['parental_education_level'] = df['parental_education_level'].fillna('Unknown')
df['distance_from_home'] = df['distance_from_home'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# Encoding Columns
# Nominal Encoding
cat_cols = ['gender','school_type','peer_influence','internet_access']
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(df[cat_cols])
new_cols = encoder.get_feature_names_out(cat_cols)
# add encoded columns to df
df[new_cols] = encoded
```

```python
df.head()
```

```python
# Ordinal Encoding
# Sort Values Orderwise (Low < Medium < High)
mapping = {'Low': 0 , 'Medium':1 , 'High': 2, 'Unknown': -1}
df['parental_involvement_encoded'] = df['parental_involvement'].map(mapping)
df['access_to_resources_encoded'] = df['access_to_resources'].map(mapping)
df['motivation_level_encoded'] = df['motivation_level'].map(mapping)
df['family_income_encoded'] = df['family_income'].map(mapping)
df['teacher_quality_encoded'] = df['teacher_quality'].map(mapping)
```

```python
df.head()
```

```python
# Binning - Numerical to Categorial
# Converting numbers to Meaningful Groups
df['study_level'] = pd.cut(df['hours_studied'] , bins = [0, 10, 20 , 45], labels =['Low', 'Medium', 'High'])
df['attendance_level'] = pd.cut(df['attendance'] , bins = [0, 70, 85, 100], labels =['Poor', 'Average', 'Good'])
df['sleep_category'] = pd.cut(df['sleep_hours'] , bins = [0, 5, 7, 10], labels =['Low Sleep', 'Normal Sleep', 'Healthy Sleep'])
df.head()
```

```python
# Scaling
scaler = StandardScaler()

scaled = scaler.fit_transform(df[['hours_studied', 'attendance', 'sleep_hours', 'previous_scores', 'tutoring_sessions']])
df[['hours_studied_scaled', 'attendance_scaled', 'sleep_hours_scaled','previous_scores_scaled', 'tutoring_sessions_scaled']] = scaled
df.head()
```

```python
# Interactive Features
df['study_effort'] = df['hours_studied'] * df['attendance']
df['learning_efficiency'] = df['previous_scores'] * df['hours_studied']
df['activity_study_balance'] = df['physical_activity'] * df['hours_studied']
df['tutoring_study_impact'] = df['tutoring_sessions'] * df['hours_studied']
```

```python
print("Shape of the Data after creating Encoding , scaling , binning and intercative features: ")
print(df.shape)
```

```python
df.to_csv('student_performance_features.csv', index=False)
```
